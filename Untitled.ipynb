```python
# Read-only OEC notebook environment check.
import os
import sys
print("python_executable:", sys.executable)
print("python_version:", sys.version.split()[0])
print("working_directory:", os.getcwd())
```

```python
# Read-only local tool checks; this cell does not access external services.
import shutil
import subprocess
for command in (["git", "--version"], [sys.executable, "-m", "pip", "--version"]):
    result = subprocess.run(command, text=True, capture_output=True, check=False, timeout=10)
    print("$", " ".join(command))
    print(result.stdout.strip() or result.stderr.strip())
    if result.returncode:
        print("TOOL_CHECK=UNAVAILABLE")
```

```python
# Read the existing Git identity without changing global configuration.
import subprocess
for key in ("user.name", "user.email"):
    result = subprocess.run(["git", "config", "--global", "--get", key],
                            text=True, capture_output=True, check=False, timeout=10)
    print(f"global_git_{key}:", result.stdout.strip() if result.returncode == 0 else "(not configured)")
print("GLOBAL_GIT_CONFIG_MUTATION=DISABLED")
```
